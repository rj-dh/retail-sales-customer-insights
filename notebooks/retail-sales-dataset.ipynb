## Calculate overall profit margin

```python
profit_margin = (df['Profit'].sum() / df['Sales'].sum()) * 100

print("Overall Profit Margin:", round(profit_margin, 2), "%")
```

## Calculate average sales per record

```python
average_sales = df['Sales'].mean()

print("Average Sales:", round(average_sales, 2))
```

## Calculate average profit

```python
average_profit = df['Profit'].mean()

print("Average Profit:", round(average_profit, 2))
```

## Calculate sales by product category

```python
df.groupby('Category')['Sales'].sum()
```

```python
df.groupby('Category')['Profit'].sum()
```

## Calculate profit margin by category

```python
category_analysis = df.groupby('Category').agg({
    'Sales': 'sum',
    'Profit': 'sum'
})
```

```python
category_analysis['Profit Margin (%)'] = (
    category_analysis['Profit'] / category_analysis['Sales']
) * 100
```

```python
category_analysis.round(2)
```

## Analyze sales by sub-category

```python
df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)
```

## Investigate discount impact on sub-category profitability

```python
df.groupby('Sub-Category').agg({
    'Discount': 'mean',
    'Sales': 'sum',
    'Profit': 'sum'
}).sort_values(by='Profit')
```

## Analyze the relationship between discount and profit

```python
df.groupby('Discount').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity': 'sum',
    'Profit': ['sum', 'mean']
})
```

## Analyze discount vs. profit

```python
discount_analysis = df.groupby('Discount').agg(
    Records=('Discount', 'size'),
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Profit=('Profit', 'mean')
)

discount_analysis.round(2)
```

```python
df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
```

```python
df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
```

## Calculate regional profit margins

```python
region_analysis = df.groupby('Region').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Records=('Region', 'count')
)

region_analysis['Profit_Margin (%)'] = (
    region_analysis['Total_Profit'] /
    region_analysis['Total_Sales'] * 100
)

region_analysis.round(2)
```

## Analyze customer segments

```python
segment_analysis = df.groupby('Segment').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Quantity_Sold=('Quantity', 'sum'),
    Records=('Segment', 'count')
)

segment_analysis.round(2)
```

## Calculate customer segment profit margins

```python
segment_analysis['Profit_Margin (%)'] = (
    segment_analysis['Total_Profit'] /
    segment_analysis['Total_Sales'] * 100
)

segment_analysis.round(2)
```

## Analyze shipping mode performance

```python
shipping_analysis = df.groupby('Ship Mode').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Quantity_Sold=('Quantity', 'sum'),
    Records=('Ship Mode', 'count')
)

shipping_analysis['Profit_Margin (%)'] = (
    shipping_analysis['Total_Profit'] /
    shipping_analysis['Total_Sales'] * 100
)

shipping_analysis.round(2)
```

## Analyze state-wise sales

```python
state_sales = df.groupby('State')['Sales'].sum().sort_values(
    ascending=False
)

state_sales.head(10)
```

## Analyze state-wise profitability

```python
state_profit = df.groupby('State')['Profit'].sum().sort_values(
    ascending=False
)

state_profit.head(10)
```

## Shipping Mode Analysis

```python
shipping_analysis = df.groupby('Ship Mode').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Quantity_Sold=('Quantity', 'sum'),
    Records=('Ship Mode', 'count')
)

shipping_analysis['Profit_Margin (%)'] = (
    shipping_analysis['Total_Profit'] /
    shipping_analysis['Total_Sales'] * 100
)

print("SHIPPING MODE ANALYSIS")
display(shipping_analysis.round(2))
```

## Top 10 States by Sales

# Create a 2x2 Business Analysis Dashboard


```python

fig, axes = plt.subplots(2, 2, figsize=(16, 11))

#chart 1: Sales by Category
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

sns.barplot(
    x=category_sales.index,
    y=category_sales.values,
    ax=axes[0, 0]
)
axes[0, 0].set_title('Total Sales by Category')
axes[0, 0].set_xlabel('Category')
axes[0, 0].set_ylabel('Sales')

#chart 2: Profit by Region
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)

sns.barplot(
    x=region_profit.index,
    y=region_profit.values,
    ax=axes[0, 1]
)
axes[0, 1].set_title('Total Profit by Region')
axes[0, 1].set_xlabel('Region')
axes[0, 1].set_ylabel('Profit')


#chart 3: Profit by Customer Segment
segment_profit = df.groupby('Segment')['Profit'].sum().sort_values(ascending=False)

sns.barplot(
    x=segment_profit.index,
    y=segment_profit.values,
    ax=axes[1, 0]
)
axes[1, 0].set_title('Total Profit by Customer Segment')
axes[1, 0].set_xlabel('Customer Segment')
axes[1, 0].set_ylabel('Profit')


#chart 4: Profit by Sub-Category
subcategory_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values()

sns.barplot(
    x=subcategory_profit.values,
    y=subcategory_profit.index,
    ax=axes[1, 1]
)
axes[1, 1].set_title('Profit by Sub-Category')
axes[1, 1].set_xlabel('Profit')
axes[1, 1].set_ylabel('Sub-Category')

plt.tight_layout()
plt.show()
```

# Sales by Region

# Profit Margin by Customer Segment


```python

plt.figure(figsize=(9, 5))

sns.barplot(
    data=segment_analysis.reset_index(),
    x='Segment',
    y='Profit_Margin (%)'
)

plt.title('Profit Margin by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Profit Margin (%)')
plt.tight_layout()
plt.show()
 
```

# Discount vs Profit


```python
plt.figure(figsize=(10, 5))

discount_profit = df.groupby('Discount')['Profit'].mean()

sns.barplot(
    x=discount_profit.index.astype(str),
    y=discount_profit.values
)

plt.title('Average Profit by Discount Level')
plt.xlabel('Discount Level')
plt.ylabel('Average Profit')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Final Retail Sales Analysis Dashboard


```python
fig, axes = plt.subplots(2, 2, figsize=(16, 11))

#1.Sales by Category
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
sns.barplot(x=category_sales.index, y=category_sales.values, ax=axes[0, 0])
axes[0, 0].set_title('Total Sales by Category')
axes[0, 0].set_ylabel('Sales')

#2.Profit by Region
region_profit = df.groupby('Region')['Profit'].sum().sort_values(ascending=False)
sns.barplot(x=region_profit.index, y=region_profit.values, ax=axes[0, 1])
axes[0, 1].set_title('Total Profit by Region')
axes[0, 1].set_ylabel('Profit')

#3.Profit by Segment
segment_profit = df.groupby('Segment')['Profit'].sum().sort_values(ascending=False)
sns.barplot(x=segment_profit.index, y=segment_profit.values, ax=axes[1, 0])
axes[1, 0].set_title('Total Profit by Customer Segment')
axes[1, 0].set_ylabel('Profit')

#4.Profit by Sub-Category
subcategory_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values()
sns.barplot(x=subcategory_profit.values, y=subcategory_profit.index, ax=axes[1, 1])
axes[1, 1].set_title('Profit by Sub-Category')
axes[1, 1].set_xlabel('Profit')

plt.suptitle('Retail Sales & Business Performance Dashboard', fontsize=16)
plt.tight_layout()

#save dashboard
plt.savefig('retail_analysis_dashboard.png', dpi=300, bbox_inches='tight')
plt.show()
```
